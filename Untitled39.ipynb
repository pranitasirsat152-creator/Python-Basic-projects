```python
df = pd.read_csv("insurance.csv")
```

```python
import pandas as pd
print(df.head(10))
```

```python
# Missing values
print(df.isnull().sum())

# Missing value percentage
print((df.isnull().sum()/len(df))*100)

# Duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df.describe()
```

```python
print("Minimum Charges:", df['charges'].min())
print("Maximum Charges:", df['charges'].max())
print("Mean Charges:", df['charges'].mean())
print("Median Charges:", df['charges'].median())
```

```python
df.hist(figsize=(12,8), bins=20)
plt.suptitle("Histogram of Numerical Features")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
encoded_df = pd.get_dummies(df, columns=['sex','smoker','region'], drop_first=True)

print(encoded_df.head())
```

```python
from sklearn.preprocessing import StandardScaler

X = encoded_df.drop('charges', axis=1)

scaler = StandardScaler()

scaled_features = scaler.fit_transform(X)

scaled_df = pd.DataFrame(scaled_features, columns=X.columns)

scaled_df.head()
```
