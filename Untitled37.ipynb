```python
import pandas as pd
df = pd.read_csv('/content/employee_salary_dataset.csv')
print(df.head(5))
```

```python
df.tail()
```

```python
df.info()
```

```python
print(df.isnull().sum())
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()
```

```python
print(df.describe())
```

```python
df.hist(figsize=(12,8))
plt.suptitle("Employee Salary Dataset Histogram")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Department', data=df)
plt.title("Department Count")
plt.xticks(rotation=45)
plt.show()

plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df)
plt.title("Gender Count")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
X = df.drop(["Monthly_Salary"], axis=1)
y = df["Monthly_Salary"]

print("Independent Features:")
print(X.columns)

print("Dependent Feature:")
print(y.name)
```

```python
encoded_df = pd.get_dummies(
    df,
    columns=["Department","Education_Level","Gender","City"],
    drop_first=True
)

print(encoded_df.head())
```

```python
scaler = StandardScaler()

encoded_df[['Experience_Years','Age']] = scaler.fit_transform(
    encoded_df[['Experience_Years','Age']]
)

print(encoded_df.head())
```
