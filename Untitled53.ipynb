```python
import pandas as pd

df = pd.read_csv("Emotion_classify_Data.csv")

print(df.head(10))
print("Shape:", df.shape)
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

print("Unique Emotions:")
print(df["Emotion"].unique())

le = LabelEncoder()
df["Emotion_Encoded"] = le.fit_transform(df["Emotion"])

mapping = dict(zip(le.classes_, le.transform(le.classes_)))

print("Mapping:")
print(mapping)

print(df.head())
```

```python
df["Comment_lower"] = df["Comment"].str.lower()

print(df[["Comment","Comment_lower"]].head())
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df["Comment_no_punct"] = df["Comment_lower"].apply(remove_punctuation)

print(df[["Comment_lower","Comment_no_punct"]].head())
```

```python
import re

def remove_numbers(text):
    return re.sub(r'\d+', '', text)

df["Comment_no_num"] = df["Comment_no_punct"].apply(remove_numbers)

print(df[["Comment_no_punct","Comment_no_num"]].head())
```

```python
def remove_special(text):
    return text.encode("ascii","ignore").decode()

df["Comment_ascii"] = df["Comment_no_num"].apply(remove_special)

print(df[["Comment_no_num","Comment_ascii"]].head())
```

```python
import nltk

nltk.download("punkt")
nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)

df["Cleaned_Text"] = df["Comment_ascii"].apply(remove_stopwords)

print(df["Cleaned_Text"].head())
```

```python
import string
import re

def clean_text(text):

    text = text.lower()

    text = text.translate(str.maketrans('', '', string.punctuation))

    text = re.sub(r'\d+', '', text)

    text = text.encode("ascii","ignore").decode()

    words = text.split()

    words = [w for w in words if w not in stop_words]

    return " ".join(words)

df["Cleaned_Text"] = df["Comment"].apply(clean_text)

print(df[["Comment","Cleaned_Text"]].head(10))
```

```python
import matplotlib.pyplot as plt

df["Text_Length"] = df["Cleaned_Text"].apply(lambda x: len(x.split()))

plt.hist(df["Text_Length"], bins=20)

plt.title("Histogram of Text Length")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")

plt.show()

print("Average:", df["Text_Length"].mean())
print("Minimum:", df["Text_Length"].min())
print("Maximum:", df["Text_Length"].max())
```

```python
# Save cleaned dataset

df.to_csv("cleaned_emotions.csv", index=False)

# Emotion Counts

print(df["Emotion"].value_counts())
```
