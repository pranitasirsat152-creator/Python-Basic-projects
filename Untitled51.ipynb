```python

# Import Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs, load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
X, y = make_blobs(
    n_samples=500,
    centers=4,
    cluster_std=0.70,
    random_state=42
)

df = pd.DataFrame(X, columns=["Feature1", "Feature2"])

print(df.head(10))
print("Shape:", df.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df)

print(pd.DataFrame(X_scaled, columns=df.columns).head())
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)

df["Cluster"] = kmeans.fit_predict(X_scaled)

print(df.head(10))
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Feature1",
    y="Feature2",
    hue="Cluster",
    data=df,
    palette="Set1"
)

plt.title("KMeans Clustering")
plt.xlabel("Feature1")
plt.ylabel("Feature2")
plt.show()
```

```python
wss = []

for k in range(1,11):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    wss.append(km.inertia_)

plt.figure(figsize=(8,5))

plt.plot(range(1,11), wss, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia (WSS)")
plt.title("Elbow Method")
plt.show()

print("Optimal K = 4")
```

```python
final_model = KMeans(n_clusters=4, random_state=42)

df["Final_Cluster"] = final_model.fit_predict(X_scaled)

print(df["Final_Cluster"].value_counts())
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Feature1",
    y="Feature2",
    hue="Final_Cluster",
    data=df,
    palette="tab10"
)

centers = scaler.inverse_transform(final_model.cluster_centers_)

plt.scatter(
    centers[:,0],
    centers[:,1],
    s=300,
    color="black",
    marker="X",
    label="Centroids"
)

plt.title("K-Means Clustering Results")
plt.legend()
plt.show()
```

```python
# Load Data
iris = load_iris()

data = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print(data.head())

# Scale
scaler = StandardScaler()
scaled = scaler.fit_transform(data)

# Elbow Method
wss = []

for i in range(1,11):
    km = KMeans(n_clusters=i, random_state=42)
    km.fit(scaled)
    wss.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1,11), wss, marker="o")
plt.title("Elbow Method")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.show()

# Final Model
model = KMeans(n_clusters=3, random_state=42)

data["Cluster"] = model.fit_predict(scaled)

# Visualization
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=data.columns[0],
    y=data.columns[1],
    hue="Cluster",
    data=data,
    palette="Set2"
)

plt.title("Final KMeans Clustering")
plt.show()

print(data.head())
print(data["Cluster"].value_counts())
```
